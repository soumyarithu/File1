```python
import numpy as np
from xgboost import XGBRegressor
import matplotlib.pyplot as plt

# Sample training data: [strength, speed, intelligence]
X = np.array([
    [10, 10, 10],  # Superman
    [6, 7, 10],    # Batman
    [9, 9, 8],     # Wonder Woman
    [8, 6, 7],     # Flash
    [5, 9, 6],     # Spider-Man
    [3, 5, 5],     # Robin
    [7, 5, 6],     # Aquaman
    [6, 6, 4],     # Black Panther
])

# Corresponding power levels out of 100
y = np.array([
    100,  # Superman
    85,   # Batman
    95,   # Wonder Woman
    90,   # Flash
    88,   # Spider-Man
    60,   # Robin
    80,   # Aquaman
    78    # Black Panther
])

# Train the model
model = XGBRegressor()
model.fit(X, y)

# User input
print("=== Superhero Power Predictor ===")
strength = int(input("Enter Strength (0–10): "))
speed = int(input("Enter Speed (0–10): "))
intelligence = int(input("Enter Intelligence (0–10): "))

# Predict power
new_hero = np.array([[strength, speed, intelligence]])
predicted_power = model.predict(new_hero)[0]

# ASCII Art
superhero_art = r"""
       __/\__
      /     o\_
     |       ___\
     |      /   o\
     |_____|_____|
    /      \     \
   /   o    \     |
  |          \   /
   \__________\/
"""

print("\n" + superhero_art)
print(f"\nPredicted Power Level: {predicted_power:.2f}/100")

# Personalized message
if predicted_power >= 95:
    status = "LEGENDARY! 🦸‍♂️🌟"
elif predicted_power >= 85:
    status = "Extremely Powerful! 💪"
elif predicted_power >= 70:
    status = "Capable Hero!"
else:
    status = "Still Training! Keep going 💥"

print(f"Status: {status}")

# Famous superhero comparison
famous_heroes = {
    "Superman": 100,
    "Batman": 85,
    "Wonder Woman": 95,
    "Flash": 90,
    "Spider-Man": 88,
    "Robin": 60,
    "Aquaman": 80,
    "Black Panther": 78
}

closest_hero = min(famous_heroes.items(), key=lambda x: abs(x[1] - predicted_power))
print(f"Closest Match: {closest_hero[0]} (Power Level: {closest_hero[1]})")

# Optional: Plot to visualize
plt.bar(["Predicted Hero", closest_hero[0]], [predicted_power, closest_hero[1]], color=["orange", "blue"])
plt.title("Superhero Power Level Comparison")
plt.ylabel("Power Level")
plt.ylim(0, 110)
plt.show()
```
